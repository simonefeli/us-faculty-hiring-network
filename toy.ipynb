# Grafo giocattolo — Blocco 1

Cinque università finte: **Alpha, Beta, Gamma, Delta, Epsilon**.
Un arco `X -> Y` con peso `w` significa: *w persone hanno preso il dottorato a X e sono state
assunte come docenti a Y*. È esattamente la stessa struttura del dataset vero, ma abbastanza
piccola da poterla calcolare a mano.

| da | a | persone |
|---|---|---:|
| Epsilon | Alpha | 2 |
| Alpha | Beta | 10 |
| Alpha | Gamma | 6 |
| Beta | Gamma | 4 |
| Gamma | Beta | 3 |
| Beta | Delta | 5 |
| Gamma | Delta | 7 |


```python
# igraph è la libreria per i grafi: la stessa usata in tutte le lezioni del corso.
import igraph

# In Python una LISTA si scrive fra parentesi quadre [ ] e separa gli elementi con la virgola.
# Ogni elemento qui dentro è una TUPLA (parentesi tonde): un gruppetto di valori in ordine fisso.
# Noi decidiamo che l'ordine sia sempre: (partenza, arrivo, quante persone).
archi = [
    ("Epsilon", "Alpha",  2),
    ("Alpha",   "Beta",  10),
    ("Alpha",   "Gamma",  6),
    ("Beta",    "Gamma",  4),
    ("Gamma",   "Beta",   3),
    ("Beta",    "Delta",  5),
    ("Gamma",   "Delta",  7),
    ("Delta",   "Alpha",  1),
]

# TupleList costruisce un grafo partendo da una lista di tuple come quella qui sopra.
#   directed=True       -> gli archi hanno un VERSO. Alpha->Beta non è la stessa cosa di Beta->Alpha.
#   edge_attrs="weight" -> il terzo valore di ogni tupla diventa un attributo dell'arco,
#                          e lo chiamiamo "weight" (peso).
# Il risultato lo mettiamo in una variabile che chiamiamo g.
g = igraph.Graph.TupleList(archi, directed=True, edge_attrs="weight")

# summary() stampa una riga di riepilogo del grafo.
print(g.summary())
```

```python
# mode="strong" -> componenti FORTEMENTE connesse: conta i versi delle frecce
print(g.connected_components(mode="weak"))
```

```python
# g.vs = the vertices. This prints their names, in igraph's internal order.
print("names       :", g.vs["name"])

# mode="in" / "out" chooses the direction. Without weights= it counts EDGES.
print("in-degree   :", g.degree(mode="in"))
print("out-degree  :", g.degree(mode="out"))

# strength() is the same thing but it SUMS THE WEIGHTS instead of counting edges.
print("in-strength :", g.strength(mode="in",  weights="weight"))
print("out-strength:", g.strength(mode="out", weights="weight"))
```
